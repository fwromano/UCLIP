```python

import sys
from pathlib import Path
from typing import Callable, List
import numpy as np
import pandas as pd
import pickle
import matplotlib.pyplot as plt
from PIL import Image

plt.rcParams.update({"figure.figsize": (8, 4), "axes.grid": True})

NOTEBOOK_ROOT = Path.cwd()
REPO_ROOT = NOTEBOOK_ROOT if (NOTEBOOK_ROOT / 'src').exists() else NOTEBOOK_ROOT.parent
SRC_ROOT = REPO_ROOT / 'src'
for path in (SRC_ROOT, REPO_ROOT):
    if str(path) not in sys.path:
        sys.path.insert(0, str(path))

print(f"Repository root: {REPO_ROOT}")
```

```python

DATA_ROOT = REPO_ROOT / 'data/car_sim/sim2_cropped_45deg'
if not DATA_ROOT.exists():
    raise FileNotFoundError(f'Missing dataset at {DATA_ROOT}')

image_paths = sorted(DATA_ROOT.glob('*/*.png'))
print(f"Found {len(image_paths)} images across {len({p.parent.name for p in image_paths})} colors.")
```

```python

def _parse_angle(path: Path) -> int:
    token = path.stem.split('_')[-1]
    try:
        return int(token)
    except ValueError:
        return -1

records = []
for path in image_paths:
    records.append({
        'image_id': f"{path.parent.name}_{path.stem.split('_')[-1]}",
        'path': path,
        'color': path.parent.name,
        'angle': _parse_angle(path)
    })

dataset_df = pd.DataFrame(records).sort_values(['color', 'angle']).reset_index(drop=True)
summary = dataset_df.groupby('color')['image_id'].count().rename('count').reset_index()
summary
```

### Selected perturbations

We inspect a subset of noise and resolution degradations applied to the same frame.

```python

TransformFn = Callable[[Image.Image], Image.Image]

def gaussian_noise_transform(std: float, seed: int) -> TransformFn:
    rng = np.random.default_rng(seed)
    def apply(image: Image.Image) -> Image.Image:
        arr = np.asarray(image).astype(np.float32) / 255.0
        noise = rng.normal(0.0, std, size=arr.shape)
        noisy = np.clip(arr + noise, 0.0, 1.0)
        return Image.fromarray((noisy * 255).round().astype(np.uint8))
    return apply

def salt_pepper_transform(amount: float, seed: int) -> TransformFn:
    rng = np.random.default_rng(seed)
    def apply(image: Image.Image) -> Image.Image:
        arr = np.asarray(image).copy()
        h, w, _ = arr.shape
        n = int(amount * h * w)
        idx = rng.choice(h * w, size=n, replace=False)
        salt = idx[: n // 2]
        pepper = idx[n // 2:]
        arr[np.unravel_index(salt, (h, w))] = 255
        arr[np.unravel_index(pepper, (h, w))] = 0
        return Image.fromarray(arr)
    return apply

def downsample_transform(percent: int, resample=Image.BICUBIC) -> TransformFn:
    scale = max(1.0 - percent / 100.0, 1e-4)
    def apply(image: Image.Image) -> Image.Image:
        w, h = image.size
        target = (max(1, int(round(w * scale))), max(1, int(round(h * scale))))
        lowres = image.resize(target, resample=resample)
        return lowres.resize((w, h), resample=resample)
    return apply

TRANSFORMS = [
    ('gaussian_0.05', 'Gaussian σ=0.05', gaussian_noise_transform(0.05, seed=1)),
    ('gaussian_0.20', 'Gaussian σ=0.20', gaussian_noise_transform(0.20, seed=2)),
    ('saltpepper_5pct', 'Salt & pepper 5%', salt_pepper_transform(0.05, seed=3)),
    ('downsample_bicubic_40pct', 'Bicubic −40%', downsample_transform(40, resample=Image.BICUBIC)),
    ('pixel_downsample_40pct', 'Pixelated −40%', downsample_transform(40, resample=Image.NEAREST)),
    ('pixel_downsample_80pct', 'Pixelated −80%', downsample_transform(80, resample=Image.NEAREST)),
]
```

```python

fig, axes = plt.subplots(1, len(TRANSFORMS) + 2, figsize=(16, 4))
axes[0].imshow(base_image)
axes[0].set_title('Original')
axes[1].imshow(base_image) if False else None
axes[1].set_title('Baseline embeddings cosines')
axes[1].axis('off')
axes[0].axis('off')
for ax, (_, label, fn) in zip(axes[2:], TRANSFORMS):
    ax.imshow(fn(base_image))
    ax.set_title(label)
    ax.axis('off')
plt.tight_layout()
plt.show()
```

### Aggregate effect across the dataset

We quantify each perturbation via mean absolute per-pixel difference (MAD) relative to the original images.

```python

def mean_absolute_difference(original: Image.Image, perturbed: Image.Image) -> float:
    a = np.asarray(original).astype(np.float32)
    b = np.asarray(perturbed).astype(np.float32)
    return float(np.mean(np.abs(a - b)))

summary_rows = []
for key, label, fn in TRANSFORMS:
    mad_values = []
    for record in records:
        with Image.open(record['path']) as img:
            original = img.convert('RGB')
            transformed = fn(original)
        mad_values.append(mean_absolute_difference(original, transformed))
    summary_rows.append({
        'transform': label,
        'category': key.split('_')[0],
        'mean_mad': np.mean(mad_values),
        'std_mad': np.std(mad_values)
    })

metric_df = pd.DataFrame(summary_rows).sort_values('mean_mad', ascending=False)
metric_df
```

```python
METRICS_PATH = REPO_ROOT / 'runs/sim2_image_perturbations/transform_metrics.pkl'
if not METRICS_PATH.exists():
    raise FileNotFoundError(f'Expected metrics at {METRICS_PATH}')

with METRICS_PATH.open('rb') as handle:
    metrics_records = pickle.load(handle)

metrics_df = pd.DataFrame(metrics_records)
transform_counts = metrics_df['transform_label'].value_counts().sort_index()
print(f"Loaded {len(metrics_df)} rows across {transform_counts.size} perturbations.")
metrics_df.head()
```

### Cosine similarity summary

Measure how deterministic and Monte Carlo mean embeddings remain aligned to the baseline across perturbations.

```python
COS_METRIC = 'mc_mean_cos_vs_baseline_mean_mean'

grouped = metrics_df.groupby(['transform_category', 'transform_label'])[COS_METRIC]
cos_summary = grouped.mean().rename('mean_cos').to_frame()
cos_summary['std_cos'] = grouped.std()
cos_summary['p05'] = grouped.quantile(0.05)
cos_summary['p95'] = grouped.quantile(0.95)
cos_summary = cos_summary.reset_index().sort_values('mean_cos', ascending=False)

cos_summary.round(4)
```

```python
plot_data = cos_summary.sort_values('mean_cos')
y_pos = np.arange(len(plot_data))
palette = {
    'reference': '#7f7f7f',
    'gaussian_noise': '#1f77b4',
    'salt_pepper': '#2ca02c',
    'downsample_bicubic': '#ff7f0e',
    'downsample_pixel': '#9467bd',
}
colors = plot_data['transform_category'].map(palette).fillna('#7f7f7f')

fig, ax = plt.subplots(figsize=(9, max(4, 0.4 * len(plot_data))))
ax.barh(y_pos, plot_data['mean_cos'], color=colors)
ax.errorbar(
    plot_data['mean_cos'],
    y_pos,
    xerr=[plot_data['mean_cos'] - plot_data['p05'], plot_data['p95'] - plot_data['mean_cos']],
    fmt='none',
    ecolor='black',
    capsize=3,
    linewidth=1,
)
ax.axvline(1.0, linestyle='--', linewidth=1, color='#333333')
ax.set_yticks(y_pos)
ax.set_yticklabels(plot_data['transform_label'])
ax.set_xlabel('cosine similarity to baseline mean')
ax.set_title('Perturbation mean embeddings vs baseline alignment')
ax.set_xlim(0.5, 1.02)
ax.grid(axis='x', linestyle=':', linewidth=0.5)
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(metric_df['transform'], metric_df['mean_mad'], color='#ff7f0e')
ax.set_ylabel('Mean absolute difference (0–255)')
ax.set_xlabel('Transform')
ax.set_title('Average pixel change per perturbation')
ax.tick_params(axis='x', labelrotation=30)
for label in ax.get_xticklabels():
    label.set_ha('right')
plt.tight_layout()
plt.show()
```

### Cosine vs noise severity

Track how increasing Gaussian noise pushes the Monte Carlo mean away from the baseline embedding.

```python
baseline_cos = float(metrics_df.loc[metrics_df['transform_key'] == 'original', COS_METRIC].mean())

noise_df = metrics_df[metrics_df['transform_category'] == 'gaussian_noise'][['severity', COS_METRIC]].copy()
noise_stats = (
    noise_df.groupby('severity')[COS_METRIC]
    .agg(mean_cos='mean', p05=lambda s: s.quantile(0.05), p95=lambda s: s.quantile(0.95))
    .reset_index()
    .sort_values('severity')
)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(noise_stats['severity'], noise_stats['mean_cos'], marker='o', color='#1f77b4', label='Gaussian noise mean')
ax.fill_between(
    noise_stats['severity'],
    noise_stats['p05'],
    noise_stats['p95'],
    color='#1f77b4',
    alpha=0.2,
    label='5th–95th percentile'
)

rng = np.random.default_rng(42)
for severity, group in noise_df.groupby('severity'):
    jitter = (rng.random(group.shape[0]) - 0.5) * 0.01
    ax.scatter(
        severity + jitter,
        group[COS_METRIC],
        color='#1f77b4',
        alpha=0.35,
        s=20,
        edgecolors='none'
    )

ax.axhline(baseline_cos, linestyle='--', color='#7f7f7f', linewidth=1, label='Baseline (original)')
ax.set_xlabel('σ')
ax.set_ylabel('cosine similarity to baseline mean')
ax.set_title('Cosine alignment under Gaussian noise')
ax.set_ylim(0.6, 1.02)
ax.set_xlim(noise_stats['severity'].min() - 0.015, noise_stats['severity'].max() + 0.015)
ax.grid(True, axis='both', linestyle=':', linewidth=0.5)
ax.legend()
plt.tight_layout()
plt.show()
```

### Cosine vs downsampling severity

Compare alignment decay for smoothed (bicubic) and pixelated crops as spatial resolution falls.

```python
baseline_cos = float(metrics_df.loc[metrics_df['transform_key'] == 'original', COS_METRIC].mean())

category_specs = {
    'downsample_bicubic': ('Bicubic (smoothed)', '#ff7f0e', -1.0),
    'downsample_pixel': ('Pixelated (nearest)', '#9467bd', 1.0),
}

summary_frames = []
for key, (label, colour, _) in category_specs.items():
    subset = metrics_df[metrics_df['transform_category'] == key]
    stats = (
        subset.groupby('severity')[COS_METRIC]
        .agg(mean_cos='mean', p05=lambda s: s.quantile(0.05), p95=lambda s: s.quantile(0.95))
        .reset_index()
        .sort_values('severity')
        .assign(category=label, colour=colour)
    )
    summary_frames.append(stats)

fig, ax = plt.subplots(figsize=(8, 4))
rng = np.random.default_rng(123)

for stats in summary_frames:
    colour = stats['colour'].iloc[0]
    label = stats['category'].iloc[0]
    key = next(k for k, v in category_specs.items() if v[0] == label)
    offset = category_specs[key][2]

    ax.plot(stats['severity'], stats['mean_cos'], marker='o', color=colour, label=label)
    ax.fill_between(stats['severity'], stats['p05'], stats['p95'], color=colour, alpha=0.2)

    subset = metrics_df[metrics_df['transform_category'] == key]
    for severity, group in subset.groupby('severity'):
        jitter = (rng.random(group.shape[0]) - 0.5) * 2.0 + offset
        ax.scatter(
            severity + jitter,
            group[COS_METRIC],
            color=colour,
            alpha=0.3,
            s=18,
            edgecolors='none'
        )

ax.axhline(baseline_cos, linestyle='--', color='#7f7f7f', linewidth=1, label='Baseline (original)')
ax.set_xlabel('Percent reduction (relative to 224px)')
ax.set_ylabel('cosine similarity to baseline mean')
ax.set_title('Cosine alignment under downsampling')
ax.set_xlim(-2, 95)
ax.set_ylim(0.6, 1.02)
ax.grid(True, axis='both', linestyle=':', linewidth=0.5)
ax.legend()
plt.tight_layout()
plt.show()
```

### Observations

- Gaussian mean embeddings stay tightly aligned (>0.95 cosine) until σ≈0.05, then drop towards ~0.72 at σ=0.50 with modest per-image scatter.
- Bicubic crops hold steady above 0.9 through 60% reduction, but once tokens shrink past ~32px the cosine falls rapidly across all samples.
- Pixelated crops diverge sooner and exhibit broader spread; many examples slip below 0.8 cosine by 80% reduction and cluster near 0.62 at 93%.
