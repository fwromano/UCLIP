## Sampling workflow

This notebook loads the dataset metadata, instruments CLIP with Monte Carlo Dropout,
collects per-transform embeddings, and writes cached summaries under `runs/sim2_image_perturbations/`.

```python
import math
import pickle
from dataclasses import dataclass
from typing import Callable, Dict, Iterable, List, Optional

import pandas as pd
import torch
from tqdm import tqdm

try:
    from uclip.core.dropout import insert_adapters, override_dropout_rate, dump_dropout_rates
    from uclip.core.sampling import sample_embeddings, compute_embedding_statistics
    from uclip.core.utils import load_clip_backbone, normalise_embeddings, set_determinism
    print('Loaded uclip core modules successfully.')
except ModuleNotFoundError as exc:
    raise ModuleNotFoundError(
        'uclip package was not found. Run the Environment bootstrap cell at the top of this notebook, restart the kernel, and re-run.'
    ) from exc
```

```python
def _parse_angle(token: str) -> Optional[int]:
    try:
        return int(token)
    except ValueError:
        return None

dataset_records: List[Dict[str, object]] = []
for path in image_paths:
    color = path.parent.name
    angle_token = path.stem.split('_')[-1]
    angle = _parse_angle(angle_token)
    dataset_records.append({
        'image_id': f"{color}_{angle_token}",
        'path': path,
        'color': color,
        'angle': angle,
    })

dataset_df = pd.DataFrame(dataset_records).sort_values(['color', 'angle']).reset_index(drop=True)
print(f'Constructed dataset dataframe with {len(dataset_df)} entries.')
dataset_df.head()
```

### CLIP configuration

Configure the CLIP backbone, dropout adapters, stochastic sampling budget, and caching locations below. Adjust `IMAGE_LIMIT` to debug on a subset before running the full sweep.

```python
MODEL_ID = 'openai/clip-vit-base-patch32'
DEVICE = None  # e.g. 'cuda' to force GPU
ADAPTER_TARGETS = [f'vision_model.encoder.layers.{idx}' for idx in range(12)] + ['visual_projection']
ADAPTER_DROPOUT_P = 0.01
OVERRIDE_DROPOUT_RATE = None  # set to a float to override existing Dropout layers
MC_PASSES = 32
MICROBATCH = 4
BASE_SEED = 20240319
IMAGE_LIMIT: Optional[int] = None  # e.g. 12 to run faster during iteration

CACHE_ROOT = REPO_ROOT / 'runs' / 'sim2_image_perturbations'
CACHE_ROOT.mkdir(parents=True, exist_ok=True)
BASELINE_CACHE_PATH = CACHE_ROOT / 'baseline_embeddings.pkl'
RESULTS_CACHE_PATH = CACHE_ROOT / 'transform_metrics.pkl'

set_determinism(BASE_SEED)
loaded = load_clip_backbone(MODEL_ID, device=DEVICE)
model, processor, device = loaded.model, loaded.processor, loaded.device
model.eval()

if ADAPTER_TARGETS:
    insert_adapters(model, ADAPTER_TARGETS, p=ADAPTER_DROPOUT_P)
if OVERRIDE_DROPOUT_RATE is not None:
    override_dropout_rate(model, OVERRIDE_DROPOUT_RATE)

drop_count, drop_rates = dump_dropout_rates(model)
print(f'Using device: {device}')
print(f'Dropout layers instrumented: {drop_count}')
if drop_rates:
    print(f"Loaded CLIP model on {device}. Gathered processor inputs successfully.")
    print('Dropout rates:', drop_rates)
```

```python
CLIP_ENCODER_BASE_PX = 224
GAUSSIAN_SIGMAS = [0.01, 0.02, 0.05, 0.10, 0.20, 0.50]
SALT_PEPPER_SETTINGS = [(0.05, 0.6)]  # (amount, salt_ratio)
DOWNSAMPLE_PERCENTS = [1, 2, 5, 10, 20, 40, 60, 80, 85, 90, 93]
NOISE_REPLICATES = 16
DOWNSAMPLE_REPLICATES = 1

@dataclass(frozen=True)
class TransformConfig:
    key: str
    label: str
    category: str
    severity: Optional[float]
    replicates: int
    factory: Callable[[int], Callable[[Image.Image], Image.Image]]

def _identity_factory(_: int) -> Callable[[Image.Image], Image.Image]:
    def apply(image: Image.Image) -> Image.Image:
        return image
    return apply

def _gaussian_factory(sigma: float) -> Callable[[int], Callable[[Image.Image], Image.Image]]:
    def make(seed: int) -> Callable[[Image.Image], Image.Image]:
        rng = np.random.default_rng(seed)
        def apply(image: Image.Image) -> Image.Image:
            array = np.asarray(image).astype(np.float32) / 255.0
            noise = rng.normal(loc=0.0, scale=sigma, size=array.shape)
            noisy = np.clip(array + noise, 0.0, 1.0)
            return Image.fromarray((noisy * 255.0).round().astype(np.uint8), mode=image.mode)
        return apply
    return make

def _salt_pepper_factory(amount: float, salt_ratio: float) -> Callable[[int], Callable[[Image.Image], Image.Image]]:
    def make(seed: int) -> Callable[[Image.Image], Image.Image]:
        rng = np.random.default_rng(seed)
        def apply(image: Image.Image) -> Image.Image:
            array = np.asarray(image).copy()
            h, w, _ = array.shape
            total = h * w
            noisy_pixels = int(amount * total)
            if noisy_pixels <= 0:
                return image
            salt_pixels = int(noisy_pixels * salt_ratio)
            pepper_pixels = noisy_pixels - salt_pixels
            indices = rng.choice(total, size=noisy_pixels, replace=False)
            salt_idx = indices[:salt_pixels]
            pepper_idx = indices[salt_pixels:]
            if salt_pixels > 0:
                ys, xs = np.divmod(salt_idx, w)
                array[ys, xs] = 255
            if pepper_pixels > 0:
                ys, xs = np.divmod(pepper_idx, w)
                array[ys, xs] = 0
            return Image.fromarray(array, mode=image.mode)
        return apply
    return make

def _downsample_factory(target_max_dim: int, resample: int) -> Callable[[int], Callable[[Image.Image], Image.Image]]:
    def make(_: int) -> Callable[[Image.Image], Image.Image]:
        def apply(image: Image.Image) -> Image.Image:
            width, height = image.size
            max_dim = max(width, height)
            if max_dim <= target_max_dim:
                return image
            scale = target_max_dim / float(max_dim)
            new_size = (max(1, int(round(width * scale))), max(1, int(round(height * scale))))
            coarse = image.resize(new_size, resample=resample)
            return coarse.resize((width, height), resample=resample)
        return apply
    return make

transform_configs: List[TransformConfig] = [
    TransformConfig(
        key='original',
        label='Original (no perturbation)',
        category='reference',
        severity=None,
        replicates=1,
        factory=_identity_factory,
    )
]

for sigma in GAUSSIAN_SIGMAS:
    transform_configs.append(
        TransformConfig(
            key=f'gaussian_sigma_{sigma:.3f}',
            label=f'Gaussian noise σ={sigma:.2f}',
            category='gaussian_noise',
            severity=sigma,
            replicates=NOISE_REPLICATES,
            factory=_gaussian_factory(sigma),
        )
    )

for amount, salt_ratio in SALT_PEPPER_SETTINGS:
    label = f'Salt & pepper {amount*100:.1f}% (salt {salt_ratio*100:.0f}%)'
    key = f'saltpepper_{int(amount*100):02d}'
    transform_configs.append(
        TransformConfig(
            key=key,
            label=label,
            category='salt_pepper',
            severity=amount,
            replicates=NOISE_REPLICATES,
            factory=_salt_pepper_factory(amount, salt_ratio),
        )
    )

for pct in DOWNSAMPLE_PERCENTS:
    target = max(1, int(round(CLIP_ENCODER_BASE_PX * (1.0 - pct / 100.0))))
    transform_configs.append(
        TransformConfig(
            key=f'downsample_{pct:02d}pct',
            label=f'Bicubic downsample {pct}% (→{target}px)',
            category='downsample_bicubic',
            severity=pct,
            replicates=DOWNSAMPLE_REPLICATES,
            factory=_downsample_factory(target, Image.BICUBIC),
        )
    )
    transform_configs.append(
        TransformConfig(
            key=f'pixel_downsample_{pct:02d}pct',
            label=f'Pixelated downsample {pct}% (→{target}px)',
            category='downsample_pixel',
            severity=pct,
            replicates=DOWNSAMPLE_REPLICATES,
            factory=_downsample_factory(target, Image.NEAREST),
        )
    )

print(f'Registered {len(transform_configs)} transform specifications')
```

```python
def encode_image(image: Image.Image) -> Dict[str, np.ndarray]:
    inputs = processor(images=image, return_tensors='pt')
    inputs = {key: value.to(device) for key, value in inputs.items()}
    with torch.no_grad():
        deterministic = model.get_image_features(**inputs)
    deterministic = normalise_embeddings(deterministic).squeeze(0)

    samples = sample_embeddings(
        model=model,
        forward_fn=model.get_image_features,
        forward_kwargs=inputs,
        passes=MC_PASSES,
        microbatch=MICROBATCH,
    )
    samples = normalise_embeddings(samples)
    stats = compute_embedding_statistics(samples)
    mc_mean = stats.mean
    mc_mean = normalise_embeddings(mc_mean.unsqueeze(0)).squeeze(0)
    cov = stats.covariance

    model.eval()  # ensure dropout modules return to eval mode

    return {
        'deterministic': deterministic.detach().cpu().numpy(),
        'mc_mean': mc_mean.detach().cpu().numpy(),
        'mc_cov_trace': float(torch.trace(cov).item()),
    }

def summarise_array(values: Iterable[float]) -> Dict[str, float]:
    arr = np.asarray(list(values), dtype=np.float64)
    if arr.size == 0:
        return {'mean': math.nan, 'std': math.nan, 'ci_low': math.nan, 'ci_high': math.nan}
    mean = float(arr.mean())
    if arr.size == 1:
        return {'mean': mean, 'std': 0.0, 'ci_low': mean, 'ci_high': mean}
    std = float(arr.std(ddof=1))
    ci = 1.96 * std / math.sqrt(arr.size)
    return {'mean': mean, 'std': std, 'ci_low': mean - ci, 'ci_high': mean + ci}

def summarise_replicates(base: Dict[str, np.ndarray], replicates: List[Dict[str, np.ndarray]]) -> Dict[str, float]:
    det_stack = np.stack([rep['deterministic'] for rep in replicates], axis=0)
    mean_stack = np.stack([rep['mc_mean'] for rep in replicates], axis=0)
    base_det = base['deterministic']
    base_mean = base['mc_mean']

    det_shift = np.linalg.norm(det_stack - base_det, axis=1)
    mc_mean_shift = np.linalg.norm(mean_stack - base_mean, axis=1)
    mc_mean_vs_det = np.linalg.norm(mean_stack - base_det, axis=1)
    cov_traces = np.asarray([rep['mc_cov_trace'] for rep in replicates], dtype=np.float64)

    det_cos_vs_base_det = det_stack @ base_det
    det_cos_vs_base_mean = det_stack @ base_mean
    mean_cos_vs_base_det = mean_stack @ base_det
    mean_cos_vs_base_mean = mean_stack @ base_mean
    baseline_det_vs_mean_cos = float(np.dot(base_det, base_mean))

    stats_det = summarise_array(det_shift)
    stats_mean = summarise_array(mc_mean_shift)
    stats_mean_vs_det = summarise_array(mc_mean_vs_det)
    stats_cov = summarise_array(cov_traces)
    stats_det_cos_det = summarise_array(det_cos_vs_base_det)
    stats_det_cos_mean = summarise_array(det_cos_vs_base_mean)
    stats_mean_cos_det = summarise_array(mean_cos_vs_base_det)
    stats_mean_cos_mean = summarise_array(mean_cos_vs_base_mean)

    if mean_stack.shape[0] > 1:
        mean_var = mean_stack.var(axis=0, ddof=1)
    else:
        mean_var = np.zeros_like(mean_stack[0])
    variance_trace = float(mean_var.sum())
    variance_mean_component = float(mean_var.mean())
    variance_max_component = float(mean_var.max())

    mean_of_means = mean_stack.mean(axis=0)
    stats = {
        'replicate_count': int(mean_stack.shape[0]),
        'deterministic_shift_l2_mean': stats_det['mean'],
        'deterministic_shift_l2_ci_low': stats_det['ci_low'],
        'deterministic_shift_l2_ci_high': stats_det['ci_high'],
        'deterministic_shift_l2_std': stats_det['std'],
        'mc_mean_shift_l2_mean': stats_mean['mean'],
        'mc_mean_shift_l2_ci_low': stats_mean['ci_low'],
        'mc_mean_shift_l2_ci_high': stats_mean['ci_high'],
        'mc_mean_shift_l2_std': stats_mean['std'],
        'mc_mean_vs_det_l2_mean': stats_mean_vs_det['mean'],
        'mc_mean_vs_det_l2_ci_low': stats_mean_vs_det['ci_low'],
        'mc_mean_vs_det_l2_ci_high': stats_mean_vs_det['ci_high'],
        'mc_mean_vs_det_l2_std': stats_mean_vs_det['std'],
        'deterministic_cos_vs_baseline_det_mean': stats_det_cos_det['mean'],
        'deterministic_cos_vs_baseline_det_ci_low': stats_det_cos_det['ci_low'],
        'deterministic_cos_vs_baseline_det_ci_high': stats_det_cos_det['ci_high'],
        'deterministic_cos_vs_baseline_det_std': stats_det_cos_det['std'],
        'deterministic_cos_vs_baseline_mean_mean': stats_det_cos_mean['mean'],
        'deterministic_cos_vs_baseline_mean_ci_low': stats_det_cos_mean['ci_low'],
        'deterministic_cos_vs_baseline_mean_ci_high': stats_det_cos_mean['ci_high'],
        'deterministic_cos_vs_baseline_mean_std': stats_det_cos_mean['std'],
        'mc_mean_cos_vs_baseline_det_mean': stats_mean_cos_det['mean'],
        'mc_mean_cos_vs_baseline_det_ci_low': stats_mean_cos_det['ci_low'],
        'mc_mean_cos_vs_baseline_det_ci_high': stats_mean_cos_det['ci_high'],
        'mc_mean_cos_vs_baseline_det_std': stats_mean_cos_det['std'],
        'mc_mean_cos_vs_baseline_mean_mean': stats_mean_cos_mean['mean'],
        'mc_mean_cos_vs_baseline_mean_ci_low': stats_mean_cos_mean['ci_low'],
        'mc_mean_cos_vs_baseline_mean_ci_high': stats_mean_cos_mean['ci_high'],
        'mc_mean_cos_vs_baseline_mean_std': stats_mean_cos_mean['std'],
        'mc_cov_trace_mean': stats_cov['mean'],
        'mc_cov_trace_ci_low': stats_cov['ci_low'],
        'mc_cov_trace_ci_high': stats_cov['ci_high'],
        'mc_cov_trace_std': stats_cov['std'],
        'mc_mean_variance_trace': variance_trace,
        'mc_mean_variance_mean_component': variance_mean_component,
        'mc_mean_variance_max_component': variance_max_component,
        'mean_shift_vs_baseline_mc_l2': float(np.linalg.norm(mean_of_means - base_mean)),
        'mean_shift_vs_baseline_det_l2': float(np.linalg.norm(mean_of_means - base_det)),
        'baseline_det_vs_mean_cos': baseline_det_vs_mean_cos,
    }
    return stats

def load_pickle(path: Path) -> Optional[object]:
    if not path.exists():
        return None
    with path.open('rb') as handle:
        return pickle.load(handle)

def save_pickle(path: Path, payload: object) -> None:
    with path.open('wb') as handle:
        pickle.dump(payload, handle)
```

```python
def compute_baseline_embeddings(force: bool = False) -> Dict[str, Dict[str, np.ndarray]]:
    cached = None if force else load_pickle(BASELINE_CACHE_PATH)
    if cached is not None:
        print(f'Loaded baseline embeddings from cache at {BASELINE_CACHE_PATH}.')
        return cached
    print('Computing baseline embeddings (dropout enabled) ...')
    baseline: Dict[str, Dict[str, np.ndarray]] = {}
    iterable = dataset_records if IMAGE_LIMIT is None else dataset_records[:IMAGE_LIMIT]
    for record in tqdm(iterable, desc='Baseline embeddings'):
        with Image.open(record['path']) as handle:
            image = handle.convert('RGB')
        encoded = encode_image(image)
        baseline[record['image_id']] = encoded
    save_pickle(BASELINE_CACHE_PATH, baseline)
    print(f'Saved baseline embeddings to {BASELINE_CACHE_PATH}.')
    return baseline

def evaluate_transforms(baseline: Dict[str, Dict[str, np.ndarray]], force: bool = False) -> List[Dict[str, object]]:
    cached = None if force else load_pickle(RESULTS_CACHE_PATH)
    results: List[Dict[str, object]] = []
    completed = set()
    if cached is not None:
        if isinstance(cached, list):
            results = list(cached)
        else:
            results = list(cached)
        completed = {entry['transform_key'] for entry in results}
        print(f'Loaded {len(results)} cached rows spanning {len(completed)} transforms from {RESULTS_CACHE_PATH}.')
    else:
        print('No cached transform metrics found; starting fresh.')

    subset = dataset_records if IMAGE_LIMIT is None else dataset_records[:IMAGE_LIMIT]
    for idx, config in enumerate(transform_configs):
        if not force and config.key in completed:
            print(f'Skipping {config.label} (cached).')
            continue

        progress = tqdm(subset, desc=config.label, leave=False, ncols=80)
        chunk: List[Dict[str, object]] = []
        for record in progress:
            base = baseline[record['image_id']]
            replicates: List[Dict[str, np.ndarray]] = []
            if config.key == 'original':
                replicates.append(base)
            else:
                for rep in range(config.replicates):
                    seed = BASE_SEED + idx * 1000 + rep
                    transform_fn = config.factory(seed)
                    with Image.open(record['path']) as handle:
                        image = handle.convert('RGB')
                    perturbed = transform_fn(image)
                    encoded = encode_image(perturbed)
                    replicates.append(encoded)
            summary = summarise_replicates(base, replicates)
            payload = {
                'image_id': record['image_id'],
                'color': record['color'],
                'angle': record['angle'],
                'transform_key': config.key,
                'transform_label': config.label,
                'transform_category': config.category,
                'severity': config.severity,
                'transform_order': idx,
                **summary,
            }
            chunk.append(payload)
        if chunk:
            results.extend(chunk)
            completed.add(config.key)
            save_pickle(RESULTS_CACHE_PATH, results)
            print(f'Checkpointed {config.label}: added {len(chunk)} rows (total {len(results)}).')
        else:
            print(f'No rows produced for {config.label}.')
    return results

BASELINE_FORCE = False
TRANSFORM_FORCE = False

print('Starting Monte Carlo evaluation pipeline ...')
baseline_embeddings = compute_baseline_embeddings(force=BASELINE_FORCE)
print(f'Baseline embeddings available for {len(baseline_embeddings)} images.')
transform_metrics = evaluate_transforms(baseline_embeddings, force=TRANSFORM_FORCE)
print(f'Computed metrics for {len(transform_metrics)} (image, transform) pairs.')
```

```python
if transform_metrics is not None:
    metrics_df = pd.DataFrame(transform_metrics)
    metrics_df = metrics_df.sort_values(['transform_order', 'color', 'angle']).reset_index(drop=True)
    print('Preview of per-image metrics dataframe:')
    display(metrics_df.head())
else:
    metrics_df = None
    print('Transform metrics dataframe is unavailable.')
```

### Grouped summaries

Aggregate the per-image metrics by colour and by viewpoint angle. Each statistic reports the mean across images along with a 95% normal approximation confidence interval (using the sample standard deviation within the group).

```python
import re
METRIC_FIELDS = {
    'deterministic_shift_l2_mean': 'Deterministic shift vs original (L2)',
    'mc_mean_shift_l2_mean': 'MCDO mean shift vs baseline mean (L2)',
    'mc_mean_vs_det_l2_mean': 'MCDO mean shift vs baseline deterministic (L2)',
    'deterministic_cos_vs_baseline_det_mean': 'Cosine(det → baseline deterministic)',
    'deterministic_cos_vs_baseline_mean_mean': 'Cosine(det → baseline MCDO mean)',
    'mc_mean_cos_vs_baseline_det_mean': 'Cosine(MCDO mean → baseline deterministic)',
    'mc_mean_cos_vs_baseline_mean_mean': 'Cosine(MCDO mean → baseline MCDO mean)',
    'mc_mean_variance_trace': 'Variance trace of replicate means',
}

def aggregate_by_group(df: pd.DataFrame, group_col: str) -> pd.DataFrame:
    records: List[Dict[str, object]] = []
    for (transform_label, severity, transform_order, group_value), chunk in df.groupby(['transform_label', 'severity', 'transform_order', group_col]):
        for field, label in METRIC_FIELDS.items():
            stats = summarise_array(chunk[field].to_numpy())
            records.append({
                'group_type': group_col,
                group_col: group_value,
                'transform_label': transform_label,
                'severity': severity,
                'transform_order': transform_order,
                'metric': label,
                'mean': stats['mean'],
                'ci_low': stats['ci_low'],
                'ci_high': stats['ci_high'],
                'n_images': chunk['image_id'].nunique(),
            })
    summary_df = pd.DataFrame(records)
    return summary_df.sort_values(['metric', group_col, 'transform_order']).reset_index(drop=True)

def _column_sort_key(name: str) -> tuple[float, float, str]:
    lower = name.lower()
    if 'original' in lower:
        group = 0
        severity = 0.0
    elif 'bicubic' in lower:
        group = 1
        match = re.search(r'([0-9]+(?:\.[0-9]+)?)', name)
        severity = float(match.group(1)) if match else 0.0
    elif 'pixel' in lower:
        group = 2
        match = re.search(r'([0-9]+(?:\.[0-9]+)?)', name)
        severity = float(match.group(1)) if match else 0.0
    elif 'gaussian' in lower:
        group = 3
        match = re.search(r'σ=([0-9]+(?:\.[0-9]+)?)', name)
        if not match:
            match = re.search(r'([0-9]+(?:\.[0-9]+)?)', name)
        severity = float(match.group(1)) if match else 0.0
    elif 'salt' in lower:
        group = 4
        match = re.search(r'([0-9]+(?:\.[0-9]+)?)', name)
        severity = float(match.group(1)) if match else 0.0
    else:
        group = 5
        severity = 0.0
    return (group, severity, name)

def pivot_summary(summary_df: pd.DataFrame, group_col: str, value_field: str = 'mean') -> Dict[str, pd.DataFrame]:
    pivots: Dict[str, pd.DataFrame] = {}
    for metric_label, group in summary_df.groupby('metric'):
        pivot = group.pivot(index=group_col, columns='transform_label', values=value_field)
        ordered_cols = sorted(pivot.columns, key=_column_sort_key)
        pivot = pivot[ordered_cols]
        pivots[metric_label] = pivot
    return pivots
```

```python
if metrics_df is not None:
    color_summary = aggregate_by_group(metrics_df, 'color')
    angle_summary = aggregate_by_group(metrics_df, 'angle')
    print('Colour-level summary (first rows):')
    display(color_summary.head())
    print('Angle-level summary (first rows):')
    display(angle_summary.head())
else:
    color_summary = angle_summary = None
    print('Skipping grouped summaries; metrics dataframe missing.')
```
